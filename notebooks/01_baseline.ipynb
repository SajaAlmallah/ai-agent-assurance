# AI Agent Assurance — Baseline

This notebook contains the first baseline experiment for the AI Agent Assurance project.

## Goal

Build a simple machine learning baseline and evaluate its performance.

```python
import pandas as pd
```

```python
from datasets import load_dataset
ds = load_dataset("neuralchemy/Prompt-injection-dataset", "core")
```

```python
train_df=ds['train'].to_pandas()
val_df=ds['validation'].to_pandas()
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer=TfidfVectorizer()
train_X=vectorizer.fit_transform(train_df['text'])
train_y=train_df['label']
val_y=val_df['label']
val_X=vectorizer.transform(val_df['text'])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
model = RandomForestClassifier(random_state=1)
model.fit(train_X,train_y)
py=model.predict(val_X)
accuracy=accuracy_score(val_y,py)
print(accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(val_y, py))
```

## Baseline Results

Model: TF-IDF + Random Forest Classifier

- Validation Accuracy: 95.2%
- Attack Precision: 93%
- Attack Recall: 99%
- Attack F1-score: 96%
- Missed Attacks: 8

This baseline provides a strong starting point for prompt injection detection.
